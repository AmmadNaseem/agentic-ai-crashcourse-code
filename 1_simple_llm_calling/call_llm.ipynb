```python
from pathlib import Path
import os

from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

# Locate the project .env whether the kernel starts here or at the repo root.
project_root = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents)
     if (folder / "pyproject.toml").is_file()),
    None,
)
if project_root is None:
    raise RuntimeError("Open this notebook from the project folder containing pyproject.toml.")

# Reload edits to .env even when this kernel already has an older value.
load_dotenv(project_root / ".env", override=True)


def require_google_api_key():
    key = os.getenv("GOOGLE_API_KEY", "").strip()
    if not key or any(marker in key.lower() for marker in
                      ("your_", "your-", "placeholder", "replace")):
        raise ValueError(
            "GOOGLE_API_KEY in the project .env is missing or still a placeholder. "
            "Create a key at https://aistudio.google.com/apikey, save it in .env, "
            "then rerun this setup cell and the Google example. "
            "A Groq key cannot authenticate Google models."
        )
    return key
```

## Call LLM Using Google LLM API

```python
llm = ChatGoogleGenerativeAI(api_key=require_google_api_key(), model="gemma-4-31b-it")

response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(api_key=require_google_api_key(), model="gemma-4-31b-it")

response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons does Jupiter have?"]
])
print(response.text)
```

### Temperature 


```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(api_key=require_google_api_key(), model="gemini-2.5-flash", temperature=0)

response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(api_key=require_google_api_key(), model="gemini-2.5-flash", temperature=1)

response = llm.invoke(prompt)
print(response.text)
```

## Call LLM Using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3-32b", reasoning_format="parsed")

messages = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human", 
        "How many moons does Jupiter have?"
    ),
]
ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
